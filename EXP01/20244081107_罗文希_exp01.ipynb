## 实验2：函数、条件、循环与递归

2.2.1. 编写score_level(score)函数：90～100分返回“优秀”，80～不足90分返回“良好”，60～不足80分返回“及格”，0～不足60分返回“不及格”。对于超出0～100范围的数值，抛出ValueError。

先检查范围，再从高到低判断等级。测试分界点、分界点以下的小数，以及两个方向的越界值。

````python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0～100 范围内")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for score in [0, 59.9, 60, 79.9, 80, 89.9, 90, 100]:
    print(f"{score}分：{score_level(score)}")
for score in [-1, 101]:
    try:
        score_level(score)
    except ValueError as error:
        print(f"{score}分：捕获 ValueError，{error}")
````

2.2.2. 分别使用for和while计算1到100的整数和，比较两种实现的结果。

`range(1, 101)` 包含 1 到 100；`while` 循环需要手动递增计数变量。

````python
for_total = 0
for number in range(1, 101):
    for_total += number

while_total = 0
number = 1
while number <= 100:
    while_total += number
    number += 1

print("for 的结果：", for_total)
print("while 的结果：", while_total)
print("结果相同：", for_total == while_total)
assert for_total == while_total == 5050
````

2.2.3. 编写递归版和迭代版阶乘函数，约定0! = 1，输入为非负整数。比较n = 0、1、5、10时的结果。

递归在 0 处终止；迭代从 1 开始累乘，n 为 0 时循环不执行。非整数抛出 `TypeError`，负整数抛出 `ValueError`。

````python
def factorial_recursive(n):
    if not isinstance(n, int) or isinstance(n, bool):
        raise TypeError("n 必须是整数")
    if n < 0:
        raise ValueError("n 必须是非负整数")
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    if not isinstance(n, int) or isinstance(n, bool):
        raise TypeError("n 必须是整数")
    if n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for number in range(1, n + 1):
        result *= number
    return result


for n in [0, 1, 5, 10]:
    recursive_result = factorial_recursive(n)
    iterative_result = factorial_iterative(n)
    print(f"{n}!：递归={recursive_result}，迭代={iterative_result}，相同={recursive_result == iterative_result}")
for function in [factorial_recursive, factorial_iterative]:
    for invalid_n in [-1, 2.5]:
        try:
            function(invalid_n)
        except (ValueError, TypeError) as error:
            print(f"{function.__name__}({invalid_n})：{type(error).__name__}，{error}")
````

2.2.4. 编写summary(values)函数，接收非空数值列表，以元组返回最小值、最大值和平均值，并通过元组解包接收结果。

返回值依次为最小值、最大值和平均值。空列表没有最小值或平均值，明确抛出 `ValueError`。

````python
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    return min(values), max(values), sum(values) / len(values)


values = [85, 58, 92, 76, 60, 89]
result = summary(values)
minimum, maximum, average = result
print("返回值及类型：", result, type(result))
print(f"解包：最小值={minimum}，最大值={maximum}，平均值={average:.2f}")
print("单个元素：", summary([60]))
print("包含负数：", summary([-3, 0, 3]))
try:
    summary([])
except ValueError as error:
    print("捕获 ValueError：", error)
````

````python
def sum_all(*args):
    return sum(args)


print("无参数：", sum_all())
print("单个参数：", sum_all(7))
print("多个参数：", sum_all(1, 2, 3.5))
````

2.4.3. 编写parse_student(line)函数：将成绩转换为整数，检查字段数量、姓名是否为空及成绩是否在0～100范围内。无效记录抛出ValueError。

依次检查字段数、去掉空白后的姓名、整数转换和成绩范围。返回包含 `name` 与 `score` 的字典。

````python
def parse_student(line):
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(fields)}")
    name = fields[0].strip()
    if not name:
        raise ValueError("姓名不能为空")
    try:
        score = int(fields[1].strip())
    except ValueError as error:
        raise ValueError(f"成绩必须是整数：{fields[1].strip()!r}") from error
    if not 0 <= score <= 100:
        raise ValueError(f"成绩必须在 0～100 范围内：{score}")
    return {"name": name, "score": score}


for line in ["Alice,85", " Zero ,0 ", "Full,100"]:
    print(repr(line), "->", parse_student(line))
for line in [",80", "Alice,-1"]:
    try:
        parse_student(line)
    except ValueError as error:
        print(repr(line), "-> 捕获 ValueError：", error)
````

2.4.4. 按行处理文件，捕获无效记录的异常，记录行号与错误原因，并继续处理后续记录。有效记录保存为字典组成的列表。

行号包含表头，因此数据从第 2 行开始。每行单独捕获 `ValueError`，失败不会中断整个文件的处理。

````python
valid_records = []
invalid_records = []
with open(students_path, "r", encoding="utf-8") as file:
    file.readline()
    for line_number, line in enumerate(file, start=2):
        try:
            record = parse_student(line)
        except ValueError as error:
            invalid_records.append({"line": line_number, "text": line.strip(), "reason": str(error)})
            print(f"第 {line_number} 行无效：{line.strip()!r}；{error}")
        else:
            valid_records.append(record)

print("有效记录：", valid_records)
print(f"有效记录数：{len(valid_records)}，无效记录数：{len(invalid_records)}")
assert len(valid_records) == 6 and len(invalid_records) == 3
assert [record["line"] for record in invalid_records] == [8, 9, 10]
````

2.5.3. 定义继承自Student的GraduateStudent类，增加research_topic属性，通过super().__init__()初始化父类属性，并重写__str__()。

子类使用父类初始化姓名与成绩，添加研究方向；重写的字符串方法保留父类信息并补充研究方向。

````python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"{super().__str__()}，研究方向：{self.research_topic}"


graduate_student = GraduateStudent("Alice", 85, "数据分析")
print(graduate_student)
print("继承的及格判断：", graduate_student.is_passed())
print("是 Student 的实例：", isinstance(graduate_student, Student))
````

2.5.4. 构造包含嵌套列表的对象，例如{"name": "Alice", "scores": [85, 90]}。分别进行直接赋值、copy.copy()和copy.deepcopy()，修改内部列表，比较结果。每次比较都从新建的原始对象开始。

直接赋值共享整个对象；浅复制的外层字典不同，但内部列表共享；深复制的内部列表也独立。每组均新建原始对象，再修改复制得到的内部列表。

````python
import copy

original = {"name": "Alice", "scores": [85, 90]}
assigned = original
assigned["scores"].append(95)
print("直接赋值：")
print("原对象：", original, "赋值对象：", assigned)
print("外层共享：", assigned is original, "内部列表共享：", assigned["scores"] is original["scores"])

original = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original)
shallow["scores"].append(95)
print("浅复制：")
print("原对象：", original, "浅复制对象：", shallow)
print("外层共享：", shallow is original, "内部列表共享：", shallow["scores"] is original["scores"])

original = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original)
deep["scores"].append(95)
print("深复制：")
print("原对象：", original, "深复制对象：", deep)
print("外层共享：", deep is original, "内部列表共享：", deep["scores"] is original["scores"])
````

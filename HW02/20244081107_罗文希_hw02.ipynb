## 字符串的乐趣
实现一些用于处理字符串的简单函数。不需要为这个问题中的任何函数执行错误检查。

1.回文是指一个词或短语，它从前往后读和从后往前读都一样。例如，level、
kayak 和 pop 都是英文回文；如果忽略空格和标点符号，rats live on no
evil star 和 Was it a car or a cat I saw? 也是回文短语。编写一个名
为 is_palindrome 的函数，它接受一个字符串作为唯一参数，并返回布尔值：如
果输入是回文，返回 True，否则返回 False。您可以假设输入仅包含英文字母
（不区分大小写）、数字 0 到 9 和空格。判断回文时应忽略空格和大小写，因此
tacocat 和 T A C O cat 都应被认为是回文。

```python
def is_palindrome(text):
    text = text.replace(" ", "").lower()
    return text == text[::-1]

print(is_palindrome("tacocat"))
print(is_palindrome("T A C O cat"))
```

2.如果一个词的字母从左到右按字母顺序排列，就称它是“字母顺序的”；允许字母重
复。例如，adder 和 beet 是字母顺序的，dog 和 cat 则不是。编写一个名为
is_abecedarian 的函数，它只接受一个字符串参数。如果参数符合上述要求，
返回 True，否则返回 False。您可以假设输入仅由字母和空格组成。判断时应忽
略空格，因此 abcd efgh xyz 被认为是字母顺序的。

```python
def is_abecedarian(text):
    text = text.replace(" ", "")
    return all(text[i] <= text[i + 1] for i in range(len(text) - 1))

print(is_abecedarian("adder"), is_abecedarian("beet"))
print(is_abecedarian("dog"), is_abecedarian("cat"))
print(is_abecedarian("abcd efgh xyz"))
```

3.编写一个名为 remove_vowels 的函数，它接受一个字符串作为唯一参数，并返回
去掉所有元音后的字符串。本题的元音是字母 a、e、i、o、u。例如，
remove_vowels('cat') 应返回 'ct'，remove_vowels('audio') 应返回
'd'。函数也应正确处理全部由元音组成的字符串。

```python
def remove_vowels(text):
    return "".join(char for char in text if char not in "aeiou")

print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(repr(remove_vowels("aeiou")))
```

## 列表的乐趣
实现一些简单的列表操作。

1.编写一个名为 list_reverse 的函数，它接受一个列表作为参数，并返回反转后的列表。例如，输入 [1, 2, 3]，应返回 [3, 2, 1]。如果输入不是列表，函数应引发适当的错误。

```python
def list_reverse(items):
    if not isinstance(items, list):
        raise TypeError("参数必须是列表")
    return items[::-1]

print(list_reverse([1, 2, 3]))
```

2.编写一个名为 is_sorted 的函数，它接受一个序列 seq 作为唯一参数。如果序列按升序排列，返回 True，否则返回 False。您可以假设 seq 是 Python 序列，且所有元素都支持比较操作，因此无需检查输入类型。函数只应遍历序列一次；需要多次遍历的低效解法不能获得满分。这个问题展示了 Python 动态类型的特点：只要 seq 支持索引，且元素支持比较，就无需预先知道序列的具体类型。

```python
def is_sorted(seq):
    for i in range(1, len(seq)):
        if seq[i] < seq[i - 1]:
            return False
    return True

print(is_sorted([1, 2, 2, 3]))
print(is_sorted((3, 1, 2)))
```

## 作为向量的元组

使用元组来表示向量。这里先用一个长度为 $d$ 的元组表示 $d$ 维向量。

1.实现一个名为 vec_scalar_mult 的函数。它接受两个参数：一个由整数或浮点数组成的元组 t，以及一个整数或浮点数 s。函数返回与 t 长度相同的元组，每个元素等于 t 的对应元素乘以 s。函数应检查输入类型；类型不合适时，引发带有适当错误消息的 TypeError。整数标量和含有整数元素的元组都是有效输入。

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("向量必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或浮点数")
    if not all(isinstance(value, (int, float)) for value in t):
        raise TypeError("向量元素必须是整数或浮点数")
    return tuple(value * s for value in t)

print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, 2, 3.5), 0.5))
```

2.实现一个名为 vec_inner_product 的函数。它接受两个由整数或浮点数组成的元组，返回这两个向量内积对应的浮点数。向量 $x,y\in\mathbb{R}^d$ 的内积是 $\sum_{j=1}^{d}x_jy_j$。若任一输入不是元组，应引发 TypeError；若两个元组的长度不一致，应引发 ValueError。

```python
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个向量必须都是元组")
    if len(x) != len(y):
        raise ValueError("两个向量的长度必须相同")
    return float(sum(a * b for a, b in zip(x, y)))

print(vec_inner_product((1, 2, 3), (4, 5, 6)))
```

3.将上述表示方法扩展到矩阵。用一个元组的元组表示 $m*n$ 矩阵 my_mx：外层元组长度为 $m$，每个内层元组长度为 $n$，表示矩阵的一行。编写一个名为 check_valid_mx 的函数，它接受一个参数；若该参数是有效矩阵则返回 True，否则返回 False。有效矩阵满足：

- 外层元组的每个元素本身都是元组；
- 所有内层元组长度相同；
- 每个内层元组的每个元素都是整数或浮点数。

```python
def check_valid_mx(my_mx):
    if not isinstance(my_mx, tuple):
        return False
    if not all(isinstance(row, tuple) for row in my_mx):
        return False
    if not my_mx:
        return True
    columns = len(my_mx[0])
    return all(
        len(row) == columns
        and all(isinstance(value, (int, float)) for value in row)
        for row in my_mx
    )

print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2), (3,))))
```

4.编写一个名为 mx_vec_mult 的函数，它接受一个矩阵（元组的元组）和一个向量（元组），返回向量与矩阵相乘得到的数值元组。函数check_valid_mx应检查参数是否合理；对于不合理的输入，应引发适当的错误。

```python
def mx_vec_mult(my_mx, vector):
    if not check_valid_mx(my_mx):
        raise ValueError("矩阵必须是各行等长、元素为数值的元组的元组")
    if not isinstance(vector, tuple):
        raise TypeError("向量必须是元组")
    if not all(isinstance(value, (int, float)) for value in vector):
        raise TypeError("向量元素必须是整数或浮点数")
    if len(vector) != len(my_mx):
        raise ValueError("行向量的长度必须等于矩阵的行数")
    # 行向量乘矩阵：分别与矩阵的每一列求内积。
    return tuple(vec_inner_product(vector, column) for column in zip(*my_mx))

print(mx_vec_mult(((1, 2, 3), (4, 5, 6)), (10, 20)))
```

4.实现 Vector.dot 方法，接收另一个 Vector 对象，返回两向量的内积。若参数类型不正确或维度不一致，应抛出合适的异常。

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        if elements is None:
            elements = (0,) * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须由列表或元组提供")
        if len(elements) != dim:
            raise ValueError("维度必须与元素个数一致")
        if not all(isinstance(value, (int, float)) for value in elements):
            raise TypeError("向量元素必须是整数或浮点数")
        # 每个向量的维度和元素可以不同，因此使用实例属性。
        self.dim = dim
        self.elements = tuple(elements)

    def _check_vector(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是Vector对象")
        if self.dim != other.dim:
            raise ValueError("两个向量的维度必须一致")

    def __eq__(self, other):
        self._check_vector(other)
        return self.elements == other.elements

    def __ne__(self, other):
        return not self == other

    def __lt__(self, other):
        self._check_vector(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check_vector(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_vector(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_vector(other)
        return self.elements >= other.elements

    def dot(self, other):
        self._check_vector(other)
        return vec_inner_product(self.elements, other.elements)

    def __mul__(self, other):
        if isinstance(other, Vector):
            self._check_vector(other)
            elements = tuple(a * b for a, b in zip(self.elements, other.elements))
        elif isinstance(other, (int, float)):
            elements = tuple(value * other for value in self.elements)
        else:
            raise TypeError("乘数必须是整数、浮点数或Vector对象")
        return Vector(self.dim, elements)

    def __rmul__(self, other):
        return self * other

v = Vector(2, (2, 3))
w = Vector(2, (4, 5))
print("默认零向量：", Vector(2).elements)
print("相等、不等、小于、小于等于、大于、大于等于：", v == w, v != w, v < w, v <= w, v > w, v >= w)
print("字典序比较：", Vector(2, (2, 4)) < Vector(2, (2, 5)))
print("内积：", v.dot(w))
print("标量左乘：", (2 * v).elements)
print("标量右乘：", (v * 2).elements)
print("逐元素乘法：", (v * w).elements)
```
